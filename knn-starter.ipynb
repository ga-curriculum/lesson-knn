## kNN is "Nonparametric"

Thus far, all of our tests and methods have been **parametric**. That is, we have assumed a certain distribution for our data:

* Linear regression (OLS) assumed a _normal distribution_ for our errors. OLS falls apart without this assumption!
* Logistic regression assume our response variable was _Bernoulli distributed_. Generally this is a safe assumption, though. We just needed to insure our observations were independent of one another.

kNN is our first **nonparametric model**. We make no strict assumptions about the shape of our data. However, this means we don't get some of the parametric niceties we're used to: coefficients, confidence intervals, etc.

## The Algorithm
The algorithm for kNN is as follows:

* Chose a positive integer $k$.
* For each observation...
    * Find the _nearest_ $k$ points to our chosen observation
    * Find the most common class among these $k$ points
        * In the case of ties, you can specify how ties are broken
    * Assign that most common class as the prediction for our observation
    
> Notes:
> * For binary classification, we prefer odd values for $k$ to avoid ties
> * You can actually also perform kNN _regression_ (instead of classification like we're doing right now). The difference is we take the _mean_ of the $k$ nearest neighbors.

 
# Distances
Remember our discussion on distance metrics. We typically pick Euclidean distance between points for kNN, but there are other options.

### Euclidean  Distance (aka $l_2$-distance)
$$ d(\mathbf{x}, \mathbf{y}) = \sqrt{\sum\limits_{i=1}^k(x_i - y_i)^2}$$

### Manhattan Distance (aka $l_1$-distance)
$$d(\mathbf{x}, \mathbf{y}) = \sum\limits_{i=1}^k \left| x_i - y_i \right|$$

### Minkowski Distance (aka $l_p$-distance)
$$d(\mathbf{x}, \mathbf{y}) = \left(\sum_{i=1}^n |x_i-y_i|^p\right)^{1/p}$$

# Conclusions and Takeaways
* k-nearest neighbors is a model that can be used for both regression and classification, but most commonly for classification
* It's a simpler model that doesn't always perform too well
* The "$k$" in kNN is what we'll come to know as a **tuning parameter** that can ge adjusted to get a better model
* kNN suffers from "the curse of dimensionality" - it gets more difficult to use and understand the more columns we have. kNN is best when our data aren't too "wide"

**When to use kNN**
* When you don't have too many rows
* When you don't have too many columns
* When you don't have any categorical features
